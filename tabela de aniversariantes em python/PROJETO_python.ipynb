```python
opcao = None
aniversarios = []
while opcao != 0:
  print("1 - ver listas dos aniversarios")
  print("2 - Adicionar novo aniversario")
  print("3 - Remover aniversario")
  print("4 - Buscar aniversario de alguem")
  print("0 - sair")
  opcao = int(input("Qual opção você quer escolher?"))
  if opcao == 0:
    print(f"Você tem {len(aniversarios)} aniversariantes em sua agenda")

  elif opcao == 1:
    print(aniversarios)

  elif opcao == 2:
    nome_adicionar = str(input("Digite o nome do aniversariante:"))
    for i in aniversarios:
      if i['nome'] == nome_adicionar:
        #leitura o nome
        while nome_adicionar == i['nome']:
          nome_adicionar = str(input("Nomes iguais não são permitidos! digite outro nome:"))
    #leitura do dia do aniversario
    dia_aniversario = int(input("Qual o dia do seu aniversario:"))
    #validando dia
    if dia_aniversario > 31:
      while dia_aniversario > 31 or dia_aniversario <= 0:
          dia_aniversario = int(input("Erro! dia do aniversario invalido digite novamente:"))
    #leitura o mês do aniversario
    mes_aniversario = int(input("Qual o mes do aniversario:"))
    #validando o mês
    if mes_aniversario > 12 or mes_aniversario <= 0:
      while mes_aniversario > 12 or mes_aniversario <= 0:
          mes_aniversario = int(input("Erro! mes do aniversario invalido digite novamente:"))
    data_adicionar = f"{dia_aniversario}/{mes_aniversario}"
    aniversarios.append({'nome' : nome_adicionar, 'data' : data_adicionar})

  elif opcao == 3:
    pessoa_excluir = str(input("Digite o nome de um aniversariante para ser retirado:"))
    for pessoa in aniversarios:
        if pessoa["nome"] == pessoa_excluir:
            aniversarios.remove(pessoa)
            print(f"Aniversariante {pessoa_excluir} removido com sucesso.")
      print("Aniversariante não existe")
  elif opcao == 4:
    buscar = str(input("digite o nome do aniversariante que você deseja buscar:"))
    for pessoa in aniversarios:
      if pessoa["nome"] == buscar:
        print(f"o aniversariante {pessoa['nome']} faz aniversario {pessoa['data']} ")
        print("Aniversariante não encontrado.")
  else:
    print("OPÇÃO INVALIDA")

  print("*"*100)

print("Lista ordenada:")
lista_ordenada = sorted(aniversarios,key=lambda pessoa : pessoa['nome'])
print(lista_ordenada)
```
